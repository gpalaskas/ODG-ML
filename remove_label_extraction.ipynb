```python
import csv
import json
import os

# Folder for the output JSON files
OUTPUT_DIR = "processed_labels"
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
removal_sample = {}
with open("annotated/combined.csv", 'r',  encoding='utf-8-sig') as f:
    reader = csv.DictReader(f, delimiter=',')
    for row in reader:
        if not row['CUI']:
            continue
        if row['CUI'] in removal_sample:
            removal_sample[row['CUI']].append(row['Remove'])
        else:
            removal_sample[row['CUI']] = [row['Remove']]
```

```python
final_sample = {key: 1 - int(value[0]) for key, value in removal_sample.items() if len(set(value)) == 1}
```

```python
with open(f'{OUTPUT_DIR}/removal_labels.json', 'w') as json_file:
    json.dump(final_sample, json_file, sort_keys=True)
```

```python
test_sample = {}
with open("annotated/test.csv", 'r',  encoding='utf-8-sig') as f:
    reader = csv.DictReader(f, delimiter=';')
    for row in reader:
        if not row['Identifier']:
            continue
        test_sample[row['Identifier']] = 1 - int(row['Remove'])
```

```python
with open(f'{OUTPUT_DIR}/test_sample.json', 'w') as json_file:
    json.dump(test_sample, json_file, sort_keys=True)
```
